Run-length encoding is a compression method that can be used to compress image files. It counts the number of consecutive pixels of each colour, then stores a code for the colour and the number of times it occurs. For example, instead of storing the code for black three times, it will store the code once, then the number 3. In some images each pixel will be a different colour. In some images all of the pixels might be the same colour. For this task, there will be no more than 9 consecutive pixels of the same colour.

For example:

Image:

| black | green | yellow |
|---|---|---|
| yellow | yellow | yellow |
| black | black | black |

Colour codes:

| Colour | Code |
|---|---|
| black | 01 |
| green | 10 |
| yellow | 11 |

This image has 9 pixels (3 × 3). Each colour in this image has a bit depth of 2, this means each colour is represented by 2 bits.

Before compression, the image will be stored as:

`011011111111010101`

When compressed the image will be stored as:

`011101114013`

**Task 4.1**

The text file `imagefile.txt` stores the colour of each pixel for an image:

- The image is 9 pixels wide by 9 pixels high
- Each colour is stored with a bit depth of 3. This means each colour is stored in 3 bits.
- Each pixel is written on a new line in the text file. For example the first line has the value 000. 000 is the colour code for the first pixel in the image.

Write program code to:

- declare a function that takes a file name as a parameter and reads in the data from the file
- declare a function that takes a list of colours as a parameter and compresses the data
- declare a function that takes the compressed data as a parameter and writes this to a text file with the name `compressedimage.txt`
- call the functions from the program with input `imagefile.txt`

The compressed data should be stored as one string of digits. **[7]**

Run your program. **[1]**


````python
#Task 4.1
#function that takes file name as parameter and reads in the data from the file
def read(filename):
    with open(filename,'r') as file:
        lines = file.readlines() #with automatically closes the file
    for i in range(len(lines)):
        lines[i] = lines[i].strip() #strip any "/n" character from the code
    return lines
#function that takes a list of colours as a parameter and compresses the data
def compress(colours):
    curr_index = 0
    compressed = ''
    flag = True #check that the curr_index is not at max limit
    while flag == True:
        count = 1 #count for consecutive codes
        compared_index = curr_index + 1 #compared index is the index after curr_index
        while True: #check the current non-consecutive colour of curr_index
            if compared_index < len(colours): #if the compared_index is not out of range
                if colours[compared_index] == colours[curr_index]: #check for same code
                    count += 1 #add consecutive count
                    compared_index += 1 #compare with the next index code
                else: #not same code (no longer consecutive)
                    compressed += (colours[curr_index]+str(count)) #add the code and the number of consecutive to the compressed string
                    curr_index = compared_index #change the curr_index to the next code
                    break
            else: #the compared_index is equal to the length of the colour list (exactly out of range), curr_index is at len of colours
                compressed += (colours[curr_index]+str(count)) #add the final code
                flag = False #trigger that we hit max index
                break #stop checking
    return compressed
#function that takes the compressed data as a parameter and writes this to a text file with the name `compressedimage.txt`
def write(compressed):
    with open('compressedimage.txt','w') as file:
        file.write(compressed+"\n")
    #with automatically closes

#main
write(compress(read('imagefile.txt')))
    
````

Write program code to:

````python
#Task 4.2
#function that takes a file name as a parameter and reads in the compressed image data from the file
def read2decom(filename):
    with open(filename,"r") as file:
        line = file.read()
        line = line.strip()
    #with automatically closes
    return line
#function that takes the data for a compressed image as a parameter and decompresses the data
def decompress(compressed):
    decompressed = []
    for i in range(0,len(compressed),4):
        code = compressed[i:i+3]
        frequency = compressed[i+3]
        for j in range(int(frequency)):
            decompressed.append(code+'\n')
    return decompressed
#function that takes the data for a decompressed image as a parameter and writes this data to a text file with the name `decompressedimage.txt`
def write2decom(decompressed):
    with open("decompressedimage.txt","w") as file:
        file.writelines(decompressed)
    #with automatically closes the file

#menu
def menu():
    filename = input("Enter the image file name: ")
    print(f"|{'Compress':^20}||{'Decompress':^20}|")
    while True:
        choice = input("Would you like to Compress(C) or Decompress(D) an image? ").lower()
        if choice in ["c","d"]:
            break
    if choice == "c":
        print(f"The compressed version of the image is: {compress(read(filename))}\nIt has been compressed to compressedimage.txt")
        write(compress(read(filename)))
    if choice == "d":
        print(f"The decompressed version of the image is: {decompress(read2decom(filename))}\nIt has been decompressed to decompressedimage.txt")
        write2decom(decompress(read2decom(filename)))

#main
menu()
````

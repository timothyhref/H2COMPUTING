````python
#Task 3.1
class Tree:
    def __init__(self,data):
        self.data = data
        self.left_pointer = None
        self.right_pointer = None
    def insert(self,data):
        if data > self.data:
            if self.right_pointer == None:
                self.right_pointer = Tree(data)
            else:
                return self.right_pointer.insert(data)
        else:
            if self.left_pointer == None:
                self.left_pointer = Tree(data)
            else:
                return self.left_pointer.insert(data)
    # def inorder(self,node): First Try, Works Too!
    #     if node:
    #         self.inorder(node.left_pointer)
    #         print(node.data)
    #         self.inorder(node.right_pointer)        
    def inorder_2(self):
        if self.left_pointer:
            self.left_pointer.inorder_2()
        print(self.data)
        if self.right_pointer:
            self.right_pointer.inorder_2()
    # def preorder(self,node): First Try, Works Too!
    #     if node:
    #         print(node.data)
    #         self.preorder(node.left_pointer)
    #         self.preorder(node.right_pointer)
    def preorder_2(self):
        print(self.data)
        if self.left_pointer:
            self.left_pointer.preorder_2()
        if self.right_pointer:
            self.right_pointer.preorder_2()

arr = [24,65,3,4,6,76,5,3,5,4]
root = Tree(arr[0])
for i in range(1,len(arr)):
    root.insert(arr[i])
root.inorder_2()
root.preorder_2()
````

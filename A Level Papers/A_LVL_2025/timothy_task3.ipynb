**Task 3.1**

Write a Python program that uses SQL to create a database with the **three** tables: `User`, `Post` and `Comment`

Define the primary and foreign keys for each table. **[6]**


````python
#Task 3.1
import sqlite3
conn = sqlite3.connect("website.db")
#create tables
conn.execute('''CREATE TABLE "User" (
	"Username"	TEXT NOT NULL,
	"Blocked"	INTEGER NOT NULL,
	PRIMARY KEY("Username")
);''')
conn.execute('''CREATE TABLE "Post" (
	"PostID"	INTEGER NOT NULL,
	"PostText"	TEXT NOT NULL,
	"DateTimePosted"	TEXT NOT NULL,
	"Username"	TEXT NOT NULL,
	FOREIGN KEY("Username") REFERENCES "User"("Username"),
	PRIMARY KEY("PostID")
);''')
conn.execute('''CREATE TABLE "Comment" (
	"CommentID"	INTEGER NOT NULL,
	"PostID"	INTEGER NOT NULL,
	"Username"	TEXT NOT NULL,
	"TheText"	TEXT NOT NULL,
	"DateTimeCommented"	TEXT NOT NULL,
	FOREIGN KEY("Username") REFERENCES "User"("Username"),
	PRIMARY KEY("CommentID"),
	FOREIGN KEY("PostID") REFERENCES "Post"("PostID")
);''')
conn.close()
````

**Task 3.2**

Write a Python program that uses SQL to insert the data from each of the **three** text files into the appropriate fields in each database table. **[4]**

Run your program. **[1]**


````python
#Task 3.2
import sqlite3
conn = sqlite3.connect('website.db')
#inserting file items into db
with open("user.txt",'r') as file:
  lines = file.readlines() #with closes automatically
for line in lines:
  line = line.strip().split(",")
  conn.execute('''INSERT INTO User VALUES(?,?)''',tuple(line))
with open("post.txt",'r') as file2:
  lines = file2.readlines()
for line in lines:
  line = line.strip().split(",")
  conn.execute('''INSERT INTO Post VALUES(?,?,?,?)''',tuple(line))
with open("comment.txt",'r') as file3:
  lines = file3.readlines()
for line in lines:
  line = line.strip().split(",")
  conn.execute('''INSERT INTO Comment VALUES(?,?,?,?,?)''',tuple(line))
conn.commit()
conn.close()
````

**Task 3.3**

Write a Python function that:

- uses SQL to search for all comments by a user who is blocked
- outputs each returned username and comment on a new line.

**[5]**

Run your program. **[1]**

Save your Jupyter Notebook for Task 3.


````python
#Task 3.3
import sqlite3
conn = sqlite3.connect("website.db")
cursor = conn.cursor()
#query joins comment and user to get user.username and comment
query = '''SELECT User.Username,Comment.TheText FROM User JOIN Comment ON User.Username = Comment.Username WHERE Blocked = 1'''
result = cursor.execute(query)
for item in result:
  print(f"The usernames are {item[0]} and the comments are {item[1]}.")
````

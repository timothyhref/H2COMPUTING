**Task 3.1**

Write the `LinkedList` class in Python. Use of a simple Python list is not sufficient. Include the following methods:

- `insert(integer_value)` inserts the `integer_value` at the beginning (head) of the list
- `delete(integer_value)` attempts to delete `integer_value` from the list; if the item was not present, return `None`
- `search(integer_value)` returns a Boolean value: `True` if `integer_value` is in the list, `False` if not in the list
- `count()` should return the number of elements in the list, or zero if empty
- `to_String()` should return a string containing a suitably formatted list with the elements separated by a comma and a space, with square brackets at either end, e.g. in the form:

  `[11, 2, 7, 4]`

  **[8]**

Test `LinkedList` by using the data in the file `Task3data.txt`. Use the `to_String()` method to print the resulting contents of the list. **[3]**


````python
# Task 3.1
class Node:
    def __init__(self,value):
        self.value = value #the value of each node
        self.next = None #the next node in the LL
class LinkedList:
    def __init__(self):
        self.head = None #no head initially
    def insert(self,integer_value):
        newnode = Node(integer_value)
        if self.head == None: #no head
            self.head = newnode #new head is new node
        else:
            newnode.next = self.head #else the next in the ll is the current head
            self.head = newnode #newnode overthrows old head
    def search(self, integer_value):
        curr = self.head
        while curr != None: #while not at end of list
            if integer_value == curr.value:
                return True #if the value exists
            else:
                curr = curr.next #otherwise continue iterating 
        return False
    def delete(self,integer_value):
        if self.search(integer_value):
            if integer_value == self.head.value:
                self.head = self.head.next #if the value wanted to delete is the head, make the next in the ll the new head
            else:
                prev = None #previous node
                curr = self.head #current node
                while curr: #while not at the end of the ll
                    if curr.value == integer_value: #if the values match
                        prev.next = curr.next #skip the current node
                        return
                    else:
                        prev = curr
                        curr = curr.next #else continue iterating
        else:
            return None #not in ll
    def count(self):
        count = 0
        curr = self.head
        while curr:
            count += 1
            curr = curr.next #iterate through, counting
        return count
    def to_String(self):
        strings = "["
        curr = self.head
        while curr:
            if curr.next != None: #not the last node
                strings += (str(curr.value) + ", ")
                curr = curr.next
            else: #the last node
                strings += (str(curr.value) + "]")
                curr = curr.next 
        return strings
#main
ll = LinkedList()
with open("Task3data.txt","r") as file:
    lines = file.readlines()
for line in lines:
    ll.insert(line.strip())
print(ll.to_String())
print(ll.count())
ll.delete("1")
print(ll.to_String())
        
````

**Task 3.2**

Write a Python subclass `SortedLinkedList` using `LinkedList` as its superclass.

The insert method in the `SortedLinkedList` subclass should ensure that the elements are stored in ascending order. **[5]**

Test `SortedLinkedList` by using the data in the file `Task3data.txt`. Use the `to_String()` method to print the resulting contents of the list.

Print the result of searching the `SortedLinkedList` for the value 94. **[2]**


````python
# Task 3.2
class SortedLinkedList(LinkedList):
    def __init__(self):
        super().__init__()
    def insert(self,integer_value):
        newnode = Node(integer_value)
        if self.head == None:
            self.head = newnode
            return
        if integer_value < self.head.value:
            newnode.next = self.head
            self.head = newnode
        else:
            prev = None
            curr = self.head
            while curr:
                if integer_value < curr.value:
                    prev.next = newnode
                    newnode.next = curr
                    return
                else:
                    prev = curr
                    curr = curr.next
            prev.next = newnode

#main
ll2 = SortedLinkedList()
with open("Task3data.txt","r") as file:
    lines = file.readlines()
for line in lines:
    ll2.insert(line.strip())
print(ll2.to_String())
````

**Task 3.3**

Write a Python subclass `Queue` using `LinkedList` as its superclass.

Additional `enqueue` and `dequeue` methods are to be defined on the `Queue` class:

- `enqueue(integer_value)` will insert `integer_value` to the end of the queue
- `dequeue()` will return the first element in the queue. If the queue is empty, return `None`. **[6]**

Test `Queue` by using the data in the file `Task3data.txt`. Print the first five elements to be dequeued from the list. **[1]**


````python
# Task 3.3
class Queue(LinkedList):
    def __init__(self):
        super().__init__()
    def enqueue(self,integer_value):
        newnode = Node(integer_value)
        if self.head == None:
            self.head = newnode
        else:
            curr = self.head
            while curr.next != None:
                curr = curr.next
            curr.next = newnode
    def dequeue(self):
        if self.head == None:
            return None
        else:
            node = self.head
            self.head = self.head.next
            return node.value

#main
queue = Queue()
with open("Task3data.txt","r") as file:
    lines = file.readlines()
for line in lines:
    queue.enqueue(line.strip())
for i in range(5):
    print(queue.dequeue())
````

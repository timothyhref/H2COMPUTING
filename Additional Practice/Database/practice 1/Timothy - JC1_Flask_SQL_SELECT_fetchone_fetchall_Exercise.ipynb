## Part 1 Set Up the Database

Run the following code to create `bookstore.db` and insert sample records.

**Do not modify this cell.**

````python
import sqlite3

conn = sqlite3.connect("bookstore.db")

conn.execute("""
CREATE TABLE IF NOT EXISTS BOOK (
    bookID INTEGER PRIMARY KEY,
    title TEXT,
    author TEXT,
    price REAL
)
""")

conn.execute("DELETE FROM BOOK")

books = [
    (101, "Python Basics", "Alice Tan", 18.90),
    (102, "Web Development with Flask", "Ben Lee", 22.50),
    (103, "Database Fundamentals", "Cheryl Lim", 20.00),
    (104, "Algorithms Made Easy", "Daniel Wong", 25.90),
    (105, "Introduction to AI", "Emma Koh", 28.00)
]

conn.executemany(
    "INSERT INTO BOOK (bookID, title, author, price) VALUES (?, ?, ?, ?)",
    books
)

conn.commit()
conn.close()

print("Database created successfully.")
````

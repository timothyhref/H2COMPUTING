## Task 3.1 Basic Flask Application

Complete `app.py`.

The home page `/` should:
1. connect to `bookstore.db`;
2. execute a `SELECT` statement to retrieve all books;
3. use `fetchall()`;
4. pass the list of books to `index.html`;
5. close the database connection.

````python
# app.py

from flask import Flask, render_template

app = Flask(__name__)

@app.route("/")
def home():
    conn = sqlite3.connect("bookstore.db")
    cursor = conn.cursor()
    cursor.execute("SELECT * FROM BOOK")
    books = cursor.fetchall()
    conn.close()
    return render_template("index.html",books = books)

if __name__ == "__main__":
    app.run(debug=True)
````

# Part 4 Search for One Book Using fetchone()

Add a second route:

```python
@app.route("/book/<int:book_id>")
```

For example:

```text
/book/103
```

should search for book ID `103`.

### Requirements
1. Get `book_id` from the URL.
2. Connect to the database.
3. Use:

```sql
SELECT * FROM BOOK WHERE bookID = ?
```

4. Use `fetchone()`.
5. If a row is found, display its details.
6. If no row is found, display:

```text
Book not found
```

### Important

Use a parameterised SQL statement:

```python
cursor = conn.execute(
    "SELECT * FROM BOOK WHERE bookID = ?",
    (book_id,)
)
```

Notice that `(book_id,)` is a **one-element tuple**.

````python
# Add this route to app.py

@app.route("/book/<int:book_id>")
def find_book(book_id):
    conn = sqlite3.connect("bookstore.db")
    cursor = conn.cursor()
    cursor.execute('''SELECT * FROM BOOK WHERE bookID = ?''',(book_id,))
    result = cursor.fetchall()
    conn.close()
    if result == []:
        return "<h1>Book not found</h1>"
    else:
        return render_template("index.html",books = result)
````

# Part 5 Understanding fetchone() and fetchall()

Complete the table.

| Situation | Return value | Python type |
|---|---|---|
| `fetchone()` finds a row | A tuple containing the field entries from that row | tuple |
| `fetchone()` finds no row | None | None Type |
| `fetchall()` finds multiple rows | List of tuples which each contain an entry | list |
| `fetchall()` finds no rows | empty list | list |

### Think About It

Why is this useful?

```python
row = cursor.fetchone()

if row is None:
    print("Book not found")
else:
    print(row)
```

Explain in your own words.

It is useful as instead of printing None, which is a vague result, it directly tells the user that the Book is not found in the database

JPBP also tags each bird in its aviary with a microchip for identification. Each bird is assigned a unique six-character `MicrochipID` consisting only of uppercase letters (`'A'` to `'Z'`) and digits (`'0'` to `'9'`), for example `"3B7K9Q"`.

To store and retrieve MicrochipIDs efficiently, JPBP uses a Binary Search Tree (BST). Each node of this BST is represented as a **list of three elements**, in the following fixed order:

```
[Left, MicrochipID, Right]
```

- `Left` (index 0) references the left child node, or `None` if there is no left child,
- `MicrochipID` (index 1) stores the six-character code of the bird's microchip,
- `Right` (index 2) references the right child node, or `None` if there is no right child.

For example, a leaf node storing MicrochipID `"3B7K9Q"` is represented as:

```
[None, "3B7K9Q", None]
```

The variable `TRoot` stores the root node of the BST. You may assume the following constants have already been defined, and may use them in your solution:

where `Status` is exactly the text `Used` or `Not Used`. For example:


## Task 2.1 &nbsp;`[3]`

Write a Python function `getNextMicrochipID(filename)` that:

1. reads the file `filename` to find the **first** MicrochipID whose status is `"Not Used"`;
2. updates that MicrochipID's status to `"Used"` in the file, so that it will not be selected again;
3. returns a BST node, represented as a list, containing the MicrochipID found in step 1, in the format `[None, MicrochipID, None]`.

If every MicrochipID in the file is already marked `"Used"`, the function should return `None`.


````python
def getNextMicrochipID(filename):
    # Write your code here
    global lines
    with open(filename,'r') as file:
        lines = file.readlines()
        lines = [line.strip().split(',') for line in lines]
   
    for line in lines:
        if line[1] == "Not Used":
            with open(filename,'w') as file:
                line[1] = "Used"
                lines = [(','.join(line)+"\n") for line in lines]
                file.writelines(lines)
            return [None, line[0], None]
        
    return None
````

## Task 2.2 &nbsp;`[6]`

Write a function `insert(TRoot, newNode)` that inserts `newNode` into the BST referenced by its root node `TRoot`, according to its `MicrochipID`.

The function should traverse the BST starting from `TRoot` and insert `newNode` at the appropriate position such that the BST property of storing data in ascending alphanumeric order is maintained, and return the root node of the updated BST.

You may assume that MicrochipIDs are always unique, so `insert` does not need to check for duplicates.

If the BST is initially empty, `newNode` should become the root node.


````python
def insert(TRoot, newNode):
    if TRoot == None:
        return newNode
    if TRoot[1] >= newNode[1]:
        TRoot[0] = insert(TRoot[0],newNode)
    else:
        TRoot[2] = insert(TRoot[2],newNode)
    return TRoot    
````

The procedure should display an appropriate output if `TRoot` is `None`.


````python
def inOrder(TRoot):
    if TRoot:
        inOrder(TRoot[0])
        print(TRoot[1],end=" ")
        inOrder(TRoot[2])
    else:
        print("You have reached the end of a branch.")
````

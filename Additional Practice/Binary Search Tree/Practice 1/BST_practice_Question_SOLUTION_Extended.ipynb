# Question 2 (Version 2): Binary Search Tree (List Representation): SOLUTION

Each BST node is a **list of three elements** `[Left, MicrochipID, Right]`, with constants `LEFT = 0`, `DATA = 1`, `RIGHT = 2`. MicrochipIDs are unique six-character codes of uppercase letters and digits. The file `microchip_ids.txt` holds lines of `MicrochipID,Status` where Status is exactly `Used` or `Not Used`.

```python
LEFT = 0
DATA = 1
RIGHT = 2
```

## Task 2.1 `[3]`

Read the file to find the **first** MicrochipID whose status is `Not Used`, mark it `Used` in the file, and return the new node `[None, MicrochipID, None]`. Return `None` if every ID is already used.

```python
def getNextMicrochipID(filename):
    # Read every line into a list so file can be rewritten afterwards
    with open(filename, 'r') as f:
        lines = f.readlines()

    found = None                                  # MicrochipID to hand out
    for i in range(len(lines)):
        # Each line is "MicrochipID,Status"; strip newline before splitting
        microchip_id, status = lines[i].strip().split(",")
        if status == "Not Used":
            found = microchip_id                  # first unused ID found
            lines[i] = microchip_id + ",Used\n"   # update its status in copy
            break                                 # stop at FIRST unused ID only

    if found is None:                             # every ID already marked Used
        return None

    # write whole file back so chosen ID cannot be selected again
    with open(filename, "w") as f:
        f.writelines(lines)

    # new node is a list: no children yet, so left and right are None
    return [None, found, None]
```

## Task 2.2 `[6]`

Insert `newNode` into the BST so ascending alphanumeric order is maintained, and return the root of the updated BST. If the BST is empty, `newNode` becomes the root.

```python
def insert(TRoot, newNode):
    if TRoot is None:   # empty tree
        return newNode  # insert as root

    current = TRoot                           # start from root
    while True:
        if newNode[DATA] < current[DATA]:     # smaller IDs go left subtree
            if current[LEFT] is None:         # insert as left leaf
                current[LEFT] = newNode
                break
            current = current[LEFT]           # else keep going left
        else:                                 # larger IDs go right subtree
            if current[RIGHT] is None:
                current[RIGHT] = newNode
                break
            current = current[RIGHT]

    return TRoot                              # root unchanged if non-empty tree
```

## Task 2.3 `[4]`

In-order traversal displays every MicrochipID in ascending alphanumeric order. An appropriate message is shown when the tree is empty.

```python
def inOrder(TRoot):
    if TRoot is None:                             # empty tree
        print("The BST is empty.")
        return

    # In-order: left subtree, current root node, right subtree.
    # Children are checked before recursing so empty-tree message
    # above is only printed for an empty tree.
    if TRoot[LEFT] is not None:
        inOrder(TRoot[LEFT])
    print(TRoot[DATA])
    if TRoot[RIGHT] is not None:
        inOrder(TRoot[RIGHT])
```

## Task 2.4 `[3]`

Main program: start with an empty BST, obtain 30 MicrochipIDs with `getNextMicrochipID()` and insert each into the BST with `insert`, updating `TRoot` each time.

```python
# Main
TRoot = None                                  # step 1: empty BST

for i in range(30):                           # step 2 and 3: 30 newly tagged birds
    newNode = getNextMicrochipID("microchip_ids.txt")
    if newNode is None:                       # no unused IDs left: stop safely
        print("No more unused MicrochipIDs available.")
        break
    TRoot = insert(TRoot, newNode)            # keep returned root

# Verification (not required by the question): display the BST contents
# in ascending order to show all 30 IDs were inserted correctly.
inOrder(TRoot)
```

## Extension: iterative in-order traversal

The recursive `inOrder` uses the call stack implicitly. The iterative version makes that stack explicit, a direct application of the stack ADT. The invariant: every node still on the stack has its own value and right subtree not yet displayed. Each node is pushed and popped exactly once, so it is O(n) time and the stack never grows beyond the height of the tree.

Common slips to avoid: the outer loop must continue while `current is not None` **or** the stack is non-empty (checking the stack alone stops before the root's right spine is visited), and the right child is reached by moving `current` to it, never by pushing it onto the stack.

```python
def inOrderIterative(TRoot):
    # An explicit stack replaces the call stack of the recursive version
    if TRoot is None:                         # empty tree
        print("The BST is empty.")
        return

    stack = []                                # nodes waiting to be displayed
    current = TRoot

    while current is not None or len(stack) > 0:
        # 1. Slide as far left as possible, stacking each node passed
        while current is not None:
            stack.append(current)             # remember the node
            current = current[LEFT]           # keep going left

        # 2. Most recently stacked node is the smallest not yet displayed
        current = stack.pop()
        print(current[DATA])

        # 3. Now visit its right subtree the same way
        current = current[RIGHT]
```

```python
# Verification: must display exactly the same output as the recursive inOrder
inOrderIterative(TRoot)
```

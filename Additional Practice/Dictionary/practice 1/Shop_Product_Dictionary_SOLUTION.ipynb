# Practical Programming Question: Python Dictionary

## Shop Product Management (10 Marks): SOLUTION

A shop stores information about its products using a Python dictionary. Each **product ID** is used as the key. The value is another dictionary containing the **product name** and **price**.

Tasks:
- **(a)** Display the product ID, product name and price of all products costing more than $40. **[3 marks]**
- **(b)** Calculate and display the total value of all the products in the dictionary. **[3 marks]**
- **(c)** Ask the user to enter a product ID. Search the dictionary and display the product name and price if the product ID exists. Otherwise, display `"Product not found"`. **[4 marks]**

Requirements: use the given dictionary, access keys and values appropriately, do not hard-code answers, display prices to 2 decimal places.


```python
# The given dictionary: product ID maps to an inner dictionary of name and price
products = {
    "P001": {"name": "Wireless Mouse", "price": 25.90},
    "P002": {"name": "Keyboard", "price": 45.50},
    "P003": {"name": "USB-C Cable", "price": 12.90},
    "P004": {"name": "Webcam", "price": 59.90},
    "P005": {"name": "Headphones", "price": 79.90}
}

# (a) Display products costing more than $40                           [3]
# .items() gives each key (product ID) together with its value (inner dictionary),
print("Products costing more than $40:")
for product_id, info in products.items():  # loop over each key-value pair
    if info["price"] > 40:                 # inner dictionary's price
        # :.2f formats the price to exactly 2 decimal places
        print(f"{product_id} {info['name']} ${info['price']:6.2f}")
```

```python
# (b) Calculate and display the total value of all products            [3]
total = 0                                  # running total
for info in products.values():             # only inner dictionaries are needed
    total = total + info["price"]          # accumulate each product price

print(f"Total value of all products: ${total:.2f}")
```

```python
# (c) Look up a product ID entered by the user                         [4]
product_id = input("Enter product ID: ")   # read the ID as a string

# in operator checks membership among dictionary KEYS,
# where product IDs are stored
if product_id in products:
    print(f"Product: {products[product_id]['name']}")
    print(f"Price: ${products[product_id]['price']:.2f}")
else:
    print("Product not found")
```

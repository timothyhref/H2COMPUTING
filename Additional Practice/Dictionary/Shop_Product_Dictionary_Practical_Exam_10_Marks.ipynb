# Practical Programming Question — Python Dictionary

## Shop Product Management — 10 Marks

A shop stores information about its products using a Python dictionary. Each **product ID** is used as the key. The value is another dictionary containing the **product name** and **price**.

You are given the following dictionary:

```python
products = {
    "P001": {"name": "Wireless Mouse", "price": 25.90},
    "P002": {"name": "Keyboard", "price": 45.50},
    "P003": {"name": "USB-C Cable", "price": 12.90},
    "P004": {"name": "Webcam", "price": 59.90},
    "P005": {"name": "Headphones", "price": 79.90}
}
```

### (a) Display the product ID, product name and price of all products costing more than $40. **[3 marks]**

### (b) Calculate and display the total value of all the products in the dictionary. **[3 marks]**

### (c) Ask the user to enter a product ID. Search the dictionary and display the product name and price if the product ID exists. Otherwise, display `"Product not found"`. **[4 marks]**

### Example output

```text
Products costing more than $40:
P002 Keyboard $45.50
P004 Webcam $59.90
P005 Headphones $79.90

Total value of all products: $224.10

Enter product ID: P003
Product: USB-C Cable
Price: $12.90
```

### Requirements
- Use the given dictionary.
- Use dictionary keys and values appropriately to access the product information.
- Do not hard-code the answers.
- Prices should be displayed to **2 decimal places**.


````python
# Write your Python program here.

products = {
    "P001": {"name": "Wireless Mouse", "price": 25.90},
    "P002": {"name": "Keyboard", "price": 45.50},
    "P003": {"name": "USB-C Cable", "price": 12.90},
    "P004": {"name": "Webcam", "price": 59.90},
    "P005": {"name": "Headphones", "price": 79.90}
}

# (a) Display products costing more than $40.
values = list(products.values())
id = list(products.keys())
cost_more = []
for i in range(len(values)):
    if values[i]["price"] > 40:
        cost_more.append(i)
print("Products costing more than $40:")
for index in cost_more:
    print(f'{id[index]:^10}{values[index]["name"]:^10}{values[index]["price"]:^10}')
# (b) Calculate and display the total value of all products.
total = 0
for item in values:
    total += item.get("price")
print(f"Total value of all products: ${total:.2f}")

# (c) Ask for a product ID and display its information or "Product not found".
query = input("Enter product ID: ")
flag = False
for i in range(len(id)):
    if id[i] == query:
        flag = True
        print(f"Product: {values[i]['name']}")
        print(f"Price: {values[i]['price']}")
if flag == False:
    print("Product not found")
````

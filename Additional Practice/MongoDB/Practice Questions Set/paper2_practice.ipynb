```python
import csv
from pymongo import *
documents = []
client = MongoClient("127.0.0.1",27017)
db = client.get_database("csv")
coll = db.get_collection("members_csv")
with open("task_files/members_import.csv","r") as file:
    rows = csv.reader(file,delimiter = ",")
    next(rows) #skips a row
    for row in rows:
        result = coll.count_documents({"member_id":row[0]})
        if result == 0:
            documents.append({"member_id":row[0],"name":row[1],"cca":row[2],"year":int(row[3]),"active":row[4]=="true","points":int(row[5]),"skills":row[6].split("|")})
        else:
            print(row[0])
coll.insert_many(documents)
```

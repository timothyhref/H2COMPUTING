## Task 2 — Add Image Information to the Database

The `Product` table contains:

| ProductID | Name | Price | Image |
|---|---|---:|---|
| P001 | Apple | 1.20 | apple.jpg |
| P002 | Banana | 0.80 | banana.jpg |
| P003 | Orange | 1.50 | orange.jpg |

Add the Image field to the database if it does not already exist.

````python
# Task 2
# Write your SQL / Python code here to add the Image field
# and/or populate the image filenames in the Product table.

import sqlite3
conn = sqlite3.connect("shop.db")
conn.execute("DROP TABLE IF EXISTS Product")
conn.execute('''CREATE TABLE IF NOT EXISTS "Product" (
	"ProductID"	TEXT,
	"Name"	TEXT,
	"Price"	NUMERIC,
	"Image"	TEXT,
	PRIMARY KEY("ProductID")
);''')

with open("Product.txt",'r') as file:
    lines = file.readlines()
    lines = [line.strip().split(',') for line in lines]
for i in range(1,len(lines)):
    conn.execute('''INSERT INTO Product VALUES (?,?,?,?)''',(lines[i]))
conn.commit()
conn.close()
````

## Task 3 — Retrieve the Product Records

````python
# Task 3
# Write your Flask route here.
from flask import *
app = Flask(__name__)
@app.route("/")
def products():
    conn = sqlite3.connect("shop.db")
    cursor = conn.cursor()
    cursor.execute("SELECT * FROM Product")
    products = cursor.fetchall()
    return render_template("products_2.html",products=products)
if __name__ == "__main__":
    app.run()
    
````

```python
#task 1.1
def task1_1(artefacts_list):
    valid = []
    for item in artefacts_list:
        artefactID = item[0]
        rating = item[1]
        flag = True
        #validation checks for id
        if artefactID[0].isalpha() == False or artefactID[0].isupper() == False or len(artefactID[1:]) != 4 or artefactID[1:].isdigit() == False or int(artefactID[1]) < 1:
            flag = False
            print('ArtefactID is invalid.')
        #validation checks for the rating
        if rating < 1 or rating > 100:
            flag = False
            print('Rating is invalid')
        if flag == True:
            valid.append(tuple([artefactID,rating]))
    return valid

#main
import ast
with open("ARTEFACTS.txt",'r') as file:
    line = ast.literal_eval(file.read()) #interpretes string as python literals
    #with autocloses
valid = task1_1(line)
print(f'The number of valid artefacts in the list are {len(valid)}.')
print(f'The first five valid artefacts in the list are {[valid[i] for i in range(5)]}')
```

```python
#task 1.2
def task1_2(artefacts_list):
    if len(artefacts_list) <= 1:
        return artefacts_list
    pivot = artefacts_list[0][1]
    less = []
    more = []
    for i in range(1,len(artefacts_list)):
        if artefacts_list[i][1] < pivot: #element is greater than the rating of the pivot
            less.append(artefacts_list[i])
        else:
            more.append(artefacts_list[i])
    return task1_2(more) + [artefacts_list[0]] + task1_2(less) #sorts list by placing elements of higher value before pivot and elements of lower value after pivot

#main
sor = task1_2(line)
print(sor)
```

```python
#task 1.3
def task1_3(artefacts_list, target_rating):
    if len(artefacts_list) == 1 and artefacts_list[0][1] == target_rating:
        return artefacts_list
    elif len(artefacts_list) <= 1:
        return None
    else:
        pivot = artefacts_list[len(artefacts_list)//2] #pivot set as mid for b-search
        left = task1_3(artefacts_list[:len(artefacts_list)//2],target_rating) #all the items greater
        right = task1_3(artefacts_list[(len(artefacts_list)//2)+1:], target_rating) #all the items smaller
        if pivot[1] == target_rating: #pivot is the searched for item
            result = []
            if left != None: #there is an item with the searched rating in the left branch
                result += left
            result.append(pivot)
            if right != None: #there is an item with the searched rating in the right branch
                result += right
            return result
        if pivot[1] < target_rating:
            return task1_3(artefacts_list[:len(artefacts_list)//2],target_rating)
        else:
            return task1_3(artefacts_list[(len(artefacts_list)//2)+1:], target_rating)

#main
result = task1_3(sor,85)
if result != None:
    print(f"The artefacts with rating 85 are: {result}")
else:
    print("There are no artefacts with rating 85")
result = task1_3(sor,100)
if result != None:
    print(f"The artefacts with rating 100 are: {result}")
else:
    print("There are no artefacts with rating 100")

```

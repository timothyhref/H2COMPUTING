```python
#task 2.5
class LinkedList:
    def __init__(self):
        self.nodes_array = [Node() for i in range(10)]
        for i in range(10):
            if i != 9:
                self.nodes_array[i].next_pointer = i+1
        self.start_pointer = -1
        self.next_free_pointer = 0
    def display_records(self):
        if self.start_pointer == -1:
            print("The linked list is empty, there are no available records.")
        else:
            curr = self.start_pointer
            while curr != -1:
                node = self.nodes_array[curr]
                print(f"Student ID: {node.get_id()}, Name: {node.get_name()}")
                curr = self.nodes_array[curr].next_pointer
    def insert_record(self,student_id,student_name):
        if self.next_free_pointer == -1:
            print("The Linked List is full.")
        else:
            index = self.next_free_pointer
            self.next_free_pointer = self.nodes_array[self.next_free_pointer].next_pointer
            self.nodes_array[index].set_id(student_id)
            self.nodes_array[index].set_name(student_name)
            self.nodes_array[index].next_pointer = -1
            if self.start_pointer == -1:
                self.start_pointer = index
            else:
                if self.nodes_array[self.start_pointer].get_id() > student_id:
                    self.nodes_array[index].next_pointer = self.start_pointer
                    self.start_pointer = index
                else:
                    curr = self.start_pointer
                    prev = None
                    while curr != -1:
                        if self.nodes_array[curr].get_id() > student_id:
                            self.nodes_array[prev].next_pointer = index
                            self.nodes_array[index].next_pointer = curr
                            break
                        else:
                            prev = curr
                            curr = self.nodes_array[curr].next_pointer
                    if curr == -1:
                        self.nodes_array[prev].next_pointer = index

    def delete_record(self,student_id):
        if self.start_pointer == -1:
            print("record not found")
        else:
            if self.nodes_array[self.start_pointer].get_id() == student_id:
                deleted_index = self.start_pointer
                self.start_pointer = self.nodes_array[deleted_index].next_pointer
                self.nodes_array[deleted_index].next_pointer = self.next_free_pointer
                self.next_free_pointer = deleted_index
            else:
                curr = self.start_pointer
                prev = None
                while curr != -1:
                    if self.nodes_array[curr].get_id() == student_id:
                        deleted_index = curr
                        self.nodes_array[prev].next_pointer = self.nodes_array[curr].next_pointer
                        self.nodes_array[deleted_index].next_pointer = self.next_free_pointer
                        self.next_free_pointer = deleted_index
                        print("record found")
                        break
                    else:
                        prev = curr
                        curr = self.nodes_array[curr].next_pointer
                if curr == -1:
                    print("record not found")

#main
ll = LinkedList()
ll.insert_record(101,"Alice")
ll.insert_record(103,"Bob")
ll.insert_record(102,"Charlie")
ll.display_records()
ll.delete_record(102)
ll.delete_record(104)
ll.display_records()
```

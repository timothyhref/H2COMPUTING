```python
#task 4.1
import sqlite3
conn = sqlite3.connect("BUSROUTES.db")
conn.execute('''CREATE TABLE IF NOT EXISTS "Service" (
	"ServiceNo"	TEXT,
	"Operator"	TEXT,
	"Category"	TEXT,
	PRIMARY KEY("ServiceNo","Operator")
);''')
conn.execute('''CREATE TABLE IF NOT EXISTS "Stop" (
	"RoadName"	TEXT,
	"Description"	TEXT,
	"BusStopCode"	INTEGER,
	PRIMARY KEY("BusStopCode","Description")
);''')
conn.execute('''CREATE TABLE IF NOT EXISTS "Route" (
	"ServiceNo"	TEXT,
	"Operator"	TEXT,
	"Direction"	INTEGER,
	"StopSequence"	INTEGER,
	"BusStopCode"	INTEGER,
	FOREIGN KEY("BusStopCode") REFERENCES "Stop"("BusStopCode"),
	FOREIGN KEY("ServiceNo") REFERENCES "Service"("ServiceNo"),
	PRIMARY KEY("ServiceNo","Operator","Direction","StopSequence"),
	FOREIGN KEY("Operator") REFERENCES "Service"("Operator")
);''')
conn.commit()
conn.close()
```

```python
#task 4.2
import sqlite3
conn = sqlite3.connect("BUSROUTES.db")
with open('Service.csv','r') as service:
    services = service.readlines()
    services = [services[i].strip().split(',') for i in range(1,len(services))]
for i in range(len(services)):
    conn.execute('''INSERT INTO Service VALUES(?,?,?)''',(services[i][0],services[i][1],services[i][2]))
with open('Stop.csv','r') as stop:
    stops = stop.readlines()
    stops = [stops[i].strip().split(',') for i in range(1,len(stops))]
for i in range(len(stops)):
    conn.execute('''INSERT INTO Stop VALUES(?,?,?)''',(stops[i][1],stops[i][2],stops[i][0]))
with open('Route.csv','r') as route:
    routes = route.readlines()
    routes = [routes[i].strip().split(',') for i in range(1,len(routes))]
for i in range(len(routes)):
    conn.execute('''INSERT INTO Route VALUES(?,?,?,?,?)''',(routes[i][0],routes[i][1],routes[i][2],routes[i][3],routes[i][4]))
conn.commit()
conn.close()
```

```python
#task 3.3
import math
class TournamentTree:
    def __init__(self,root=None):
        self.root = None
        self.team_code = ord("A")-1
        self.match_count = 1
    def create_tree(self,num_of_teams):
        depth = math.log(num_of_teams,2)
        if depth == 0:
            self.team_code += 1
            return Node("Team "+chr(self.team_code))

        left_ptr = self.create_tree(2**(depth-1))
        right_ptr = self.create_tree(2**(depth-1))
        node = Node("Match "+str(self.match_count))
        self.match_count += 1
        node.left_ptr = left_ptr
        node.right_ptr = right_ptr
        self.root = node
        return node
    def level_order(self):
        '''
        print tree from root
        print out all the nodes of the same level in a row 
        '''
        self._level_order(self.root)
    
    def _level_order(self, node_list):
        '''
        Helper function for internal use
        print tree from node list
        '''
        # convert node_list to a list if it is not
        if not isinstance(node_list, list):
            node_list = [node_list]
        # Stop recursion if the list is empty
        if not node_list:
            return
        # define a list to collect node in next layer
        next_layer = []
        while node_list:
            node = node_list.pop(0) # Remove the first item
            print(node, end = ' ')
            if node.left_ptr:
                next_layer.append(node.left_ptr)
            if node.right_ptr:
                next_layer.append(node.right_ptr)
        print() # print new line after processing a layer
        self._level_order(next_layer)

#main 
tree = TournamentTree()
tree.create_tree(8)
tree.level_order()
```

```python
#task 3.5
import math
class TournamentTree:
    def __init__(self,root=None):
        self.root = None
        self.team_code = ord("A")-1
        self.match_count = 1
    def create_tree(self,num_of_teams):
        depth = math.log(num_of_teams,2)
        if depth == 0:
            self.team_code += 1
            return Node("Team "+chr(self.team_code))

        left_ptr = self.create_tree(2**(depth-1))
        right_ptr = self.create_tree(2**(depth-1))
        node = Node("Match "+str(self.match_count))
        self.match_count += 1
        node.left_ptr = left_ptr
        node.right_ptr = right_ptr
        self.root = node
        return node
    def level_order(self):
        '''
        print tree from root
        print out all the nodes of the same level in a row 
        '''
        self._level_order(self.root)
    
    def _level_order(self, node_list):
        '''
        Helper function for internal use
        print tree from node list
        '''
        # convert node_list to a list if it is not
        if not isinstance(node_list, list):
            node_list = [node_list]
        # Stop recursion if the list is empty
        if not node_list:
            return
        # define a list to collect node in next layer
        next_layer = []
        while node_list:
            node = node_list.pop(0) # Remove the first item
            print(node, end = ' ')
            if node.left_ptr:
                next_layer.append(node.left_ptr)
            if node.right_ptr:
                next_layer.append(node.right_ptr)
        print() # print new line after processing a layer
        self._level_order(next_layer)

    def pre_order(self,node):
        if node:
            print(node)
            self.pre_order(node.left_ptr)
            self.pre_order(node.right_ptr)
    def post_order(self,node):
        if node:
            self.post_order(node.left_ptr)
            self.post_order(node.right_ptr)
            print(node)
    def search(self,target_name,node):
        if node:
            if node.team == target_name:
                return True
            else:
                left = self.search(target_name,node.left_ptr)
                right = self.search(target_name,node.right_ptr)
                if not (left or right):
                    return False
                else:
                    return True
#main
tree = TournamentTree()
tree.create_tree(8)
print(tree.search("Team E",tree.root))
print(tree.search("Team J", tree.root))
```

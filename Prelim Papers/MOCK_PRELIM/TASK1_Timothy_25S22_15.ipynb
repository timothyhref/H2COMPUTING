Banks have reported a sharp rise in phishing text messages that imitate parcel deliveries and bank alerts. A telecommunications company is building a spam filter that uses machine learning: instead of following fixed rules written by a programmer, the filter improves its performance as more labelled example messages are collected.

Each message has already been converted into two numeric features:

- `links` – the number of web links the message contains
- `urgent` – the number of urgency words the message contains (for example 'now', 'immediately', 'suspended').

The text file `TRAINING.txt` contains labelled example messages, one per line, in the format:

`links,urgent,label`

where `label` is either `PHISHING` or `GENUINE`.

The text file `TESTDATA.txt` contains further labelled messages, in the same format, that were held back for evaluating the model.

The text file `UNKNOWN.txt` contains new messages to classify, one per line, in the format:

`messageID,links,urgent`

The filter uses the k-nearest neighbours (k-NN) algorithm, an example of supervised learning: a new message is given the label held by the majority of its k nearest labelled examples.

**Task 1.1**

Write program code to read the training data from `TRAINING.txt` into an appropriate data structure.

Output the number of `PHISHING` examples and the number of `GENUINE` examples in the training data. **[4]**


````python
#Task 1.1
with open('TRAINING.txt','r') as file:
    lines = file.readlines()
#with automatically closes
lines = [line.strip().split(',') for line in lines]
phishing = 0
genuine = 0
for line in lines:
    if line[2] == "GENUINE":
        genuine += 1 #count if genuine
    else:
        phishing += 1 #count if phishing
print(f"Number of Phishing examples: {phishing}")
print(f"Number of Genuine examples: {genuine}")
````

**Task 1.2**

The distance between two messages is the Euclidean distance between their features. For two messages with features (`links1`, `urgent1`) and (`links2`, `urgent2`):

distance = √((links1 − links2)² + (urgent1 − urgent2)²)

Write a function `task1_2(links1, urgent1, links2, urgent2)` that returns this distance.

Test your function with the two messages (3, 5) and (0, 1).

`task1_2(3, 5, 0, 1)` should return and display `5.0` **[4]**


````python
#Task 1.2
import math
def task1_2(links1,urgent1,links2,urgent2):
    distance = math.sqrt((links1 - links2)**2 + (urgent1 - urgent2)**2) #uses formula to calculate the distance
    return distance
print(task1_2(3,5,0,1))
````

**Task 1.3**

Write a function `task1_3(links, urgent, k)` that:

- calculates the distance from the message (`links`, `urgent`) to every message in the training data
- returns a list of the labels of the `k` nearest training messages.

Test your function by outputting the result of `task1_3(3, 3, 3)` **[6]**


````python
#Task 1.3
def task1_3(links,urgent,k):
    import copy
    distances = [task1_2(links,urgent,int(line[0]),int(line[1])) for line in lines]
    lines_copy = copy.deepcopy(lines)
    for i in range(len(distances)):
        lines_copy[i].append(distances[i]) #now labels goes up to index 3 where it is the distance
    #insertion sort
    for i in range(1,len(lines_copy)):
        curr = lines_copy[i]
        j = i-1
        while j>=0 and lines_copy[j][3] > curr[3]:
            lines_copy[j+1] = lines_copy[j]
            j-=1
        lines_copy[j+1] = curr
    return [lines_copy[i][2] for i in range(k)]

print(task1_3(3,3,3))
````

**Task 1.4**

Write a function `task1_4(links, urgent, k)` that:

- uses your function from Task 1.3
- returns the label held by the majority of the `k` nearest training messages. `k` is always an odd number.

Use your function, with `k = 3`, to classify every message in `UNKNOWN.txt`. Output each message ID and its label. **[6]**


````python
#Task 1.4
def task1_4(lines,urgent,k):
    labels = task1_3(lines,urgent,k)
    num_phish = labels.count("PHISHING")
    num_genuine = labels.count("GENUINE")
    if num_phish>num_genuine:
        return "PHISHING"
    else:
        return "GENUINE"
with open("UNKNOWN.txt",'r') as unclassed:
    messages = unclassed.readlines()
    messages = [line.strip().split(',') for line in messages]
    #with auto closes
for message in messages:
    label = task1_4(int(message[1]),int(message[2]),3)
    print(f"Message ID: {message[0]}. Label: {label}.")
````

**Task 1.5**

The value of `k` is a parameter of the model that can be tuned.

Write program code to evaluate the model. For each of the values `k` = 1, 3 and 5:

- classify every message in `TESTDATA.txt`
- compare the label produced by the model with the correct label in the file
- output the value of `k` and the accuracy (the number of correct classifications out of the total).

State, in a comment or a markdown cell, which value of `k` performed best on the test data and suggest a reason. **[4]**


````python
#Task 1.5
import copy
with open("TESTDATA.txt",'r') as test:
    messages = test.readlines()
    messages = [line.strip().split(',') for line in messages]
    #with auto closes 
for i in range(1,6,2):
    valid = 0
    for message in messages:
        label = task1_4(int(message[0]),int(message[1]),i)
        if label == message[2]:
            valid += 1
    print(f"For a k value of {i}, the accuracy is {valid} over {len(messages)} total messages")
    
````

Since it compares against the most number of messages, it is the most accurate in prediction since we account for greater uncertainty, e.g. anomolous messages that don't follow the trend.

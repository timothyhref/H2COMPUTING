- `CheckSum("chat-mini", 0)` should return and display `34` **[7]**


````python
#Task 3.2
def CheckSum(Name,Total):
    if len(Name) > 1:
        Total += (ord(Name[0]))*len(Name)
        NewName = Name[1:]
        return CheckSum(NewName,Total)
    else:
        Total += ord(Name)
        return Total % 97
print(CheckSum("vision-small", 0))
print(CheckSum("chat-mini", 0))
````

````python
#Task 3.3
def task3_3(name, low, high):
    if low > high:
        return -1
    mid = (high + low)//2
    if catalogue[mid][0] == name:
        return catalogue[mid][1]
    else:
        if catalogue[mid][0] > name:
            high = mid - 1
        else:
            low = mid + 1
        return task3_3(name,low,high)
    
with open('MODELS.txt','r') as file:
    catalogue = file.readlines()
    catalogue = [line.strip().split(",") for line in catalogue]
    #with auto closes()
check = CheckSum("vision-ultra", 0)
real = int(task3_3("vision-ultra",0,len(catalogue)-1))
if check == real:
    print("AUTHENTIC")
elif real == -1:
    print("Model not in catalogue")
else:
    print("TAMPERED - do not use")
````

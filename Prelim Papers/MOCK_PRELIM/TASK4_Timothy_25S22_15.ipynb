**Task 4.1**

Write a Python program that uses SQL to create a database with two tables: one table to store data about the students and one table to store data about the tutoring sessions.

Define the primary and foreign keys for each table. **[6]**


````python
#Task 4.1
import sqlite3
conn = sqlite3.connect("tutoring.db")
conn.execute('''CREATE TABLE IF NOT EXISTS "Students" (
	"studentID"	INTEGER NOT NULL,
	"name"	TEXT NOT NULL,
	"className"	TEXT NOT NULL,
	PRIMARY KEY("studentID")
);''')
conn.execute('''CREATE TABLE IF NOT EXISTS "Sessions" (
	"sessionID"	INTEGER NOT NULL,
	"studentID"	INTEGER NOT NULL,
	"date"	TEXT NOT NULL,
	"questionsAsked"	TEXT NOT NULL,
	"topicArea"	TEXT NOT NULL,
	"rating"	INTEGER NOT NULL,
	FOREIGN KEY("studentID") REFERENCES "Students"("studentID"),
	PRIMARY KEY("sessionID")
);''')
conn.commit()
conn.close()
````

**Task 4.2**

Write a Python program that uses SQL to insert the data from each of the two text files into the appropriate fields in each database table. **[4]**

Run your program to test the data has been entered correctly. **[1]**


````python
#Task 4.2
import sqlite3
conn = sqlite3.connect("tutoring.db")
with open("students.txt",'r') as file:
    lines = file.readlines()
    lines = [line.strip().split(',') for line in lines]
for line in lines:
    conn.execute('''INSERT INTO Students VALUES (?,?,?)''',(line))
with open("sessions.txt",'r') as file2:
    lines = file2.readlines()
    lines = [line.strip().split(',') for line in lines]
for line in lines:
    conn.execute('''INSERT INTO Sessions VALUES (?,?,?,?,?,?)''',(line))
conn.commit()
conn.close()
````

**Task 4.3**

The school wants a usage report.

Write a Python program that uses SQL to show, for each student who has used the assistant at least once:

- the student's name
- the total number of questions they have asked
- their average rating of the sessions (to 2 decimal places).

The results should be shown in descending order of the total number of questions. **[6]**


````python
#Task 4.3
import sqlite3
conn = sqlite3.connect("tutoring.db")
cursor = conn.cursor()
cursor.execute('''SELECT Students.name, SUM(Sessions.questionsAsked) as total_asked, AVG(sessions.rating) as avg_rating
FROM Students JOIN Sessions ON Students.studentID = Sessions.studentID
GROUP BY Students.studentID ORDER BY total_asked Desc''')
result = cursor.fetchall()
for item in result:
    print(item)
````

**Task 4.4**

Write a Python function to:

- take as input a topic area and a month (for example, `Python` and `Mar`)
- use SQL to find the sessions for that topic area in that month
- output the number of sessions and the average rating for those sessions
- output an appropriate message if there were no sessions for that topic area in that month.

Test your program with the following data:

```
Topic area   Python
Month        Mar
```

Then test it again with a topic area and month that has no sessions. **[8]**


````python
#Task 4.4
def relevance(topic,month):
    conn = sqlite3.connect('tutoring.db')
    cursor = conn.cursor()
    search_month = f'%-{month}'
    cursor.execute('''select count(studentID), avg(rating)
    from Sessions where date like (?) AND topicArea = (?)''',(search_month,topic))
    result = cursor.fetchall()
    if result[0][0] == 0:
        print("There were no sessons for that topic area in that month")
    else:
        print(f"The total number of sessions for the topic area of {topic} in the month of {month} was {result[0][0]}\nThe average rating for those sessions was {result[0][1]}")

relevance("Python","Mar")
relevance("Pseudocode",'May')
````

**Task 2.1**

Write a class `Node` with the attributes:

- `data` – the job ID stored in the node
- `next` – a reference to the next node in the list, or `None`.

Write a class `JobList` that uses `Node` objects. Use of a simple Python list is not sufficient. Include the following methods:

- `insert(job_id)` inserts the `job_id` at the beginning (head) of the list
- `search(job_id)` returns a Boolean value: `True` if `job_id` is in the list, `False` if not
- `count()` returns the number of jobs in the list, or zero if empty
- `to_string()` returns a string containing the job IDs separated by a comma and a space, with square brackets at either end, for example:

`[J104, J103, J102, J101]` **[8]**


````python
#Task 2.1
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
class JobList:
    def __init__(self):
        self.head = None
    def insert(self,job_id):
        newnode = Node(job_id)
        if self.head == None:
            self.head = newnode
        else:
            newnode.next = self.head
            self.head = newnode
    def search(self,job_id):
        curr = self.head
        while True:
            if curr.data == job_id:
                return True
            elif curr.next == None:
                return False
            else:
                curr = curr.next
    def count(self):
        count = 0
        curr = self.head
        while curr != None:
            count+=1
            curr=curr.next
        return count
    def to_string(self):
        result = "["
        curr = self.head
        while curr.next != None:
            result += (curr.data + ", ")
            curr = curr.next
        result += (curr.data + "]")
        return result
````

**Task 2.3**

Write a subclass `JobQueue` using `JobList` as its superclass.

Additional `enqueue` and `dequeue` methods are to be defined on the `JobQueue` class:

- `enqueue(job_id)` inserts `job_id` at the end of the queue
- `dequeue()` removes and returns the first job in the queue. If the queue is empty, return `None`.

Test `JobQueue` by enqueuing all ten jobs in `JOBS.txt`, then dequeuing and printing the first **five** jobs to be sent to the GPU. Use `to_string()` to print the jobs still waiting. **[6]**


````python
#Task 2.3
class JobQueue(JobList):
    def __init__(self):
        super().__init__()
    def enqueue(self,job_id):
        newnode = Node(job_id)
        if self.head == None:
            self.head = newnode
        else:
            curr = self.head
            while curr.next != None:
                curr = curr.next
            curr.next = newnode
    def dequeue(self):
        if self.head == None:
            return None
        else:
            item = self.head
            self.head = self.head.next
            return item.data

#main
queue = JobQueue()
for line in lines:
    queue.enqueue(line)
for i in range(5):
    job = queue.dequeue()
    print(job)
print(f"Jobs still waiting: {queue.to_string()}")

        
````

Test by cancelling `J108` and then attempting to cancel `J102` (which has already been dequeued). Use `to_string()` to print the queue after each call. **[2]**


````python
#Task 2.5
class JobQueue(JobList):
    def __init__(self):
        super().__init__()
    def enqueue(self,job_id):
        newnode = Node(job_id)
        if self.head == None:
            self.head = newnode
        else:
            curr = self.head
            while curr.next != None:
                curr = curr.next
            curr.next = newnode
    def dequeue(self):
        if self.head == None:
            return None
        else:
            item = self.head
            self.head = self.head.next
            return item.data
    def print_reverse(self,node):
        if node == None:
            return
        self.print_reverse(node.next)
        print(node.data)
    def cancel(self,job_id):
        if self.head == None:
            return False
        elif self.head.data == job_id:
            self.head = None
            return True
        else:
            curr = self.head.next
            prev = self.head
            while curr != None:
                if curr.data == job_id:
                    prev.next = curr.next
                    return True
                else:
                    prev = curr
                    curr = curr.next
            return False
#main
queue = JobQueue()
for line in lines:
    queue.enqueue(line)
for i in range(5):
    job = queue.dequeue()
    print(job)
print(f"Jobs still waiting: {queue.to_string()}")
queue.cancel("J108")
print(f"Jobs still waiting: {queue.to_string()}")
queue.cancel("J102")
print(f"Jobs still waiting: {queue.to_string()}")
````

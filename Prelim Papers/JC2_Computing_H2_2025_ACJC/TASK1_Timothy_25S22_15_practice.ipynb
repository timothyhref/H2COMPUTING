```python
#Task 1.1
def check_magic(arr):
    #check 1 all appear exactly once:
    appeared = []
    for i in range(len(arr)):
        for j in range(len(arr)):
            if arr[i][j] in appeared:
                return False
            appeared.append(arr[i][j])
    #check 2, check rows
    sum_row = sum(arr[0])
    for i in range(1,len(arr)):
        if sum(arr[i]) != sum_row:
            return False
    #check 3, check diagonals:
    for col in range(len(arr)):
        sum_col = 0
        for row in range (len(arr)):
            sum_col += arr[row][col] 
        if sum_col != sum_row:
            return False
    #check 4, diagonals
    lr_coor=[0,0]
    sum_lrdia = 0
    while lr_coor != [len(arr),len(arr)]:
        sum_lrdia += arr[lr_coor[0]][lr_coor[1]]
        lr_coor[0],lr_coor[1] = lr_coor[0]+1,lr_coor[1]+1
    rl_coor=[0,len(arr)-1]
    sum_rldia = 0
    while rl_coor != [len(arr),-1]:
        sum_rldia += arr[rl_coor[0]][rl_coor[1]]
        rl_coor[0],rl_coor[1] = rl_coor[0]+1,rl_coor[1]-1
    if sum_rldia != sum_row or sum_lrdia != sum_row:
        return False
    return True
        

        
```

```python
#task1.2
def check_file(filename):
    with open(filename,'r') as file:
        lines = [line.strip().split(',') for line in file.readlines()]
        for line in lines:
            for i in range(len(lines)):
                line[i] = int(line[i])
        
        print(lines)
        #with closes automatically
        return check_magic(lines)

print(check_file("Task1-1.csv"))
print(check_file("Task1-2.csv"))
```

```python
#task1.3
def make_magic(n):
    arr = [[""]*n for i in range(n)]
    first = (n//2)+1
    arr[0][first] = 1
    prev_coor = [0,first]
    for i in range(2,(n**2)+1):
        y = (prev_coor[0]-1)
        x = (prev_coor[1]+1)%n
        if y == -1:
            y = n-1
        if arr[y][x] != '':
            y = (prev_coor[0]+1)%n
            x = prev_coor[1]

        arr[y][x] = i
        prev_coor = [y,x]
    return arr

square = [(",".join(map(str,line))+"\n") for line in make_magic(9)] #map applies the function across all items in the line
with open('Task1-3.csv','w') as file:
    file.writelines(square)
        
```

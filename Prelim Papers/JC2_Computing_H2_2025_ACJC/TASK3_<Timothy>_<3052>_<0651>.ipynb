```python
#task 3.1
class Array:
    def __init__(self,n):
        self.container = [['',None] for i in range(n)]
        for row in range(n-1):
            self.container[row][1] = (row + 1)
        self.free = 0
    def get_data(self,index):
        return self.container[index][0]
    def set_data(self,index, new_data):
        self.container[index][0] = new_data
    def get_pointer(self,index):
        return self.container[index][1]
    def set_pointer(self,index, new_pointer):
        self.container[index][1] = new_pointer
```

```python
#task 3.2
class LinkedList:
    def __init__(self,array):
        self.array = array
        self.head = None
    def display(self):
        if self.head == None:
            print("The ll is empty.")
        else:
            curr = self.head
            while curr != None:
                print(self.array.get_data(curr))
                curr = self.array.get_pointer(curr)
    def add(self,new_data):
        if self.array.free == None:
            print("The ll is full.")
        else:
            index = self.array.free #stores the index which new_data is inserted 
            self.array.set_data(index,new_data) #insert the data
            #moves the free pointer to the next free space
            self.array.free = self.array.get_pointer(self.array.free)
            self.array.set_pointer(index,None) #set next pointer to None
            
            if self.head == None: #empty ll
                self.head = index
                
            else:
                if new_data < self.array.get_data(self.head):
                    self.array.set_pointer(index,self.head) #set the pointer of the inserted node to the head
                    self.head = index
                else:
                    curr = self.head
                    prev = None
                    while curr != None:
                        if new_data < self.array.get_data(curr):
                            self.array.set_pointer(index,curr)
                            self.array.set_pointer(prev,index)
                            break
                        else:
                            prev = curr
                            curr = self.array.get_pointer(curr)
                    if curr == None:
                        self.array.set_pointer(prev,index)
    def delete(self,to_delete):
        if self.head == None:
            print("The ll is empty.")
        else:
            if self.array.get_data(self.head) == to_delete:
                index = self.head #stores the index of the deleted node
                self.head = self.array.get_pointer(self.head)
                self.array.set_pointer(index,self.array.free)
                self.array.free = index
            else:
                curr = self.head
                prev = None
                while curr != None:
                    if self.array.get_data(curr) == to_delete:
                        
                        self.array.set_pointer(prev,self.array.get_pointer(curr))
                        self.array.set_pointer(curr,self.array.free)
                        self.array.free = curr
                        break
                    else:
                        prev = curr
                        curr = self.array.get_pointer(curr)
                if curr == None:
                    print("Data to be deleted cannot be found in the linked list.")

        
                    
            
```

```python
#task 3.2
MyArray = Array(10)
AnimalList = LinkedList(MyArray)
PlantList = LinkedList(MyArray)
AnimalList.add("CAT")
PlantList.add("ROSE")
AnimalList.add("GOAT")
PlantList.add("FERN")
AnimalList.add("HORSE")
PlantList.add("VINE")
AnimalList.delete("COW")
PlantList.delete("ROSE")
print("AnimalList Contents:")
AnimalList.display()
print("PlantList Contents")
PlantList.display()
```

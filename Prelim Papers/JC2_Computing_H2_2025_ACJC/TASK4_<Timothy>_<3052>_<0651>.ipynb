```python
#task 4.1
import sqlite3
conn = sqlite3.connect("COURSES.db")
conn.execute('''CREATE TABLE IF NOT EXISTS "Instructor" (
	"InstructorID"	INTEGER,
	"Name"	TEXT,
	PRIMARY KEY("InstructorID")
);''')
conn.execute('''CREATE TABLE IF NOT EXISTS "Customer" (
	"CustomerID"	INTEGER,
	"Name"	TEXT,
	PRIMARY KEY("CustomerID")
);''')
conn.execute('''CREATE TABLE IF NOT EXISTS "Course" (
	"CourseID"	TEXT,
	"Name"	TEXT,
	"InstructorID"	INTEGER,
	"StartDate"	INTEGER,
	"Price"	INTEGER,
	PRIMARY KEY("CourseID"),
	FOREIGN KEY("InstructorID") REFERENCES "Instructor"("InstructorID")
);''')
conn.execute('''CREATE TABLE IF NOT EXISTS "Registration" (
	"CustomerID"	INTEGER,
	"CourseID"	INTEGER,
	"Paid"	INTEGER,
	FOREIGN KEY("CourseID") REFERENCES "Course"("CourseID"),
	FOREIGN KEY("CustomerID") REFERENCES "Customer"("CustomerID"),
	PRIMARY KEY("CustomerID","CourseID")
);''')
conn.commit()
conn.close()
```

```python
#task 4.2
import sqlite3
conn = sqlite3.connect("COURSES.db")
with open('COURSE.txt','r') as course:
    courses = course.readlines()
    courses = [line.strip().split(',') for line in courses]
    #with automatically closes
for i in range(len(courses)):
    conn.execute('''INSERT INTO Course VALUES (?,?,?,?,?)''',(courses[i][0],courses[i][1],courses[i][2],courses[i][3],courses[i][4]))
with open('CUSTOMER.txt','r') as customer:
    customers = customer.readlines()
    customers = [line.strip().split(',') for line in customers]
    #with automatically closes
for i in range(len(customers)):
    conn.execute('''INSERT INTO Customer VALUES (?,?)''',(customers[i][0],customers[i][1]))
with open('INSTRUCTOR.txt','r') as instructor:
    instructors = instructor.readlines()
    instructors = [line.strip().split(',') for line in instructors]
    #with automatically closes
for i in range(len(instructors)):
    conn.execute('''INSERT INTO Instructor VALUES (?,?)''',(instructors[i][0],instructors[i][1]))
with open('REGISTRATION.txt','r') as registration:
    registrations = registration.readlines()
    registrations = [line.strip().split(',') for line in registrations]
    #with automatically closes
for i in range(len(registrations)):
    conn.execute('''INSERT INTO Registration VALUES (?,?,?)''',(registrations[i][0],registrations[i][1],registrations[i][2]))
conn.commit()
conn.close()
```

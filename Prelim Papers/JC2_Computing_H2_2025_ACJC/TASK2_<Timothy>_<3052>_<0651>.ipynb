```python
#task 2.1
def quicksort(lis):
    quicksort_helper(lis,0,len(lis)-1)

def quicksort_helper(values,low,high):
    if high > low: #base case, at least 2 elements in the recursed list
        pivot = partition(values,low,high)
        quicksort_helper(values,pivot+1,high)
        quicksort_helper(values,low,pivot-1)

def partition(values,low,high):
    #set pivot as last element
    pivot = values[high]
    i = low-1 #i is the border between the values smaller and greater than the pivot
    for j in range(low,high):
        if values[j] < pivot:
            i+=1 #increase border up 1
            #swap into smaller section
            values[i],values[j] = values[j],values[i]
        #swap pivot to after the border
    values[high],values[i+1] = values[i+1],values[high]
    return i+1 #return pivot index

def insertionsort(lis):
    for i in range(1,len(lis)):
        j = i-1
        curr = lis[i]
        while j >= 0 and lis[j] > curr:
            lis[j+1] = lis[j]
            j -= 1
        lis[j+1] = curr

array = [23,4,6,2,4,76,8,5,5,7,5]
insertionsort(array)
print(array)



```

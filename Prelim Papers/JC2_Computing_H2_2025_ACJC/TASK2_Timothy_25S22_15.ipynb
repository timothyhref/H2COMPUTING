```python
#Task 2.1
def quicksort(lis):
    sort(lis)
def partition(lis,high,low):
    pivot = lis[high]
    i = low-1
    for j in range(low,high):
        if lis[j] <= pivot:
            i+=1
            lis[i],lis[j] = lis[j],lis[i]
    lis[i+1],lis[high] = lis[high],lis[i+1]
    return i+1

def sort(lis,high=-1,low=-1):
        if high == -1:
            high = len(lis)-1
        if low == -1:
            low = 0
        if high > low:
            pivot_index = partition(lis,high,low)
            sort(lis,pivot_index-1,low)
            sort(lis,high,pivot_index+1)

def insertionsort(lis):
    for i in range(1,len(lis)):
        j = i-1
        curr = lis[i]
        while lis[j] > curr and j>=0:
            lis[j+1] = lis[j]
            j-=1
        lis[j+1] = curr
    


array = [3,4,2,6,78,9,11,3,5]
quicksort(array)
print(array)
array1 = [3,4,2,6,78,9,11,3,5]
insertionsort(array1)
print(array1)
```

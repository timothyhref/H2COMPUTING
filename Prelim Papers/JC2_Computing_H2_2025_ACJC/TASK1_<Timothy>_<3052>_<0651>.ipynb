```python
#task 1.1
def check_magic(arr):
    #check for unique integers\
    present = [] #array of the integers storing the integers already present in the magic square
    for row in range(len(arr)):
        for col in range(len(arr)):
            if arr[row][col] not in present:
                present.append(arr[row][col])
            else:
                print('not unique')
                return False
    for row in range(len(arr)):
        arr[row] = list(map(int,arr[row])) #turns all the items in each row into an integer for easy calculation
    #check for the sum of rows
    sum_row = sum(arr[0])
    for row in range(1,len(arr)):
        sum_of_row = sum(arr[row])
        if sum_of_row != sum_row:
            print("row error")
            return False
    #check col
    for col in range(len(arr)):
        sum_of_col = 0
        for row in range(len(arr)):
            sum_of_col += arr[row][col]
        if sum_of_col != sum_row:
            print("col error")
            return False
    #check diagonal
    #top left to bottom right
    sum_of_dia = 0
    for i in range(len(arr)):
        sum_of_dia += arr[i][i]
    if sum_of_dia != sum_row:
        print("l-r dia error")
        return False
    #top right to bottom left
    coord = len(arr)-1 #stores the position of the column coordinate
    sum_of_dia = 0
    for i in range(len(arr)):
        sum_of_dia += arr[i][coord]
        coord -= 1
    if sum_of_dia != sum_row:
        print("r-l dia error")
        return False
    return True
```

```python
#task 1.2
def check_file(filename):
    with open(filename,'r') as file:
        lines = file.readlines()
        lines = [line.strip().split(',') for line in lines] #seperates each line into a list forming a nxn array
        #with automatically closes the file
    if check_magic(lines) == True:
        print(f"The nxn array in {filename} forms a magic square.")
    else:
        print(f"The nxn array in {filename} does not form a magic square.")

#main
check_file("Task1-1.csv")
check_file("Task1-2.csv")
```

```python
#task 1.3
def make_magic(n):
    if n%2 == 0:
        print("Unable to construct magic square of even number n")
    else:
        arr = [['']*n for i in range(n)]
        count = 2
        #start with 1
        arr[0][n//2] = 1
        coord = [0,n//2] #saves the coord of the last inserted item
        while count != (n**2 + 1):
            if (coord[0] - 1) < 0:
                new_row = n-1 #wrap around from top to bottom
            else:
                new_row = coord[0] - 1
            new_col = (coord[1] + 1)%n #wrap from right to left
            if arr[new_row][new_col] != '':
                new_row = (coord[0] + 1)%n #wrap around from bottom to top
                new_col = coord[1] 
            arr[new_row][new_col] = count
            coord[0],coord[1] = new_row,new_col
            count+=1
        return arr

#main
array = make_magic(9)
for row in range(len(array)):
    array[row] = list(map(str,array[row]))
with open('Task1-3.csv','w') as file:
    for row in range(len(array)):
        row = ','.join(array[row]) + '\n'
        file.write(row)
    #with auto closes
```

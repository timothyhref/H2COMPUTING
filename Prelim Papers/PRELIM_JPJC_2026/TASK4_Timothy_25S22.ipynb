```python
#Task 4.1
import sqlite3
conn = sqlite3.connect('HOUSEALLOCATION.db')
#create HOUSE table
conn.execute('''CREATE TABLE IF NOT EXISTS `HOUSE` (
	`houseName`	TEXT,
	`imageURL`	TEXT,
	PRIMARY KEY(`houseName`)
);''')
#create STAFF table
conn.execute('''CREATE TABLE IF NOT EXISTS `STAFF` (
	`staffID`	TEXT,
	`name`	TEXT,
	`department`	TEXT,
	`houseName`	TEXT,
	PRIMARY KEY(`staffID`),
	FOREIGN KEY(`houseName`) REFERENCES `HOUSE`(`houseName`)
);''')
conn.commit()
conn.close()
```

```python
#task 4.2
import sqlite3
conn = sqlite3.connect('HOUSEALLOCATION.db')
#insert HOUSE.txt values
with open("HOUSE.txt",'r') as house:
    records = house.readlines()
    records = [record.strip().split(',') for record in records] #list comprehension to split the values in the list
    #with auto closes
for record in records:
    conn.execute('''INSERT INTO HOUSE VALUES (?,?)''',(record))

#insert STAFF.txt values
with open("STAFF.txt",'r') as staff:
    records = staff.readlines()
    records = [record.strip().split(',') for record in records] #list comprehension to split the values in the list
    #with auto closes
for record in records:
    conn.execute('''INSERT INTO STAFF VALUES (?,?,?,?)''',(record))

conn.commit()
conn.close()
    
```

```python
#task1.1
def task1_1(filename):
    with open(filename,'r') as file: #open the file to read, with auto closes once done
        lines = file.readlines()
        lines = [line.strip().split(',') for line in lines] #list comprehension to split the fields from each line
    for i in range(len(lines)):
        lines[i] = tuple([int(lines[i][0]),str(lines[i][1]),float(lines[i][2]),int(lines[i][3])]) #assign each field its specific data type
    return lines

#main
result = task1_1("TASK1.txt")
print(result)
```

```python
#task 1.2
def task1_2(list_of_records):
    #bubble sort
    n = len(list_of_records)
    for i in range(n-1): #count passes
        for j in range(n-1-i): #count comparisons
            #ascending order
            if list_of_records[j][2] > list_of_records[j+1][2]:
                list_of_records[j],list_of_records[j+1] = list_of_records[j+1],list_of_records[j]
    return list_of_records

#main
print(task1_2(task1_1("TASK1.txt")))
```

```python
#task 1.3
def task1_3(list_of_records):
    if len(list_of_records) <= 1:
        return list_of_records
    else:
        mid = len(list_of_records)//2
        pivot = list_of_records[mid][0]
        more_than = [] #list containing elements more than the pivot
        less_than = [] #list containing elements less than the pivot
        for item in (list_of_records[:mid]+list_of_records[mid+1:]):
            if item[0] > pivot: #check each element more than pivot else is less
                more_than.append(item)
            else:
                less_than.append(item)
        return task1_3(less_than) + [list_of_records[mid]] + task1_3(more_than) #recursive call
    
#main
print(task1_3(task1_1("TASK1.txt")))
```

```python
#task 1.4
def task1_4(list_of_records,search_id,high,low):
    search_id = int(search_id)
    if low>high:
        return "Not Found" #no items left in list
    else:
        mid = (low+high)//2
        if list_of_records[mid][0] == search_id:
            return (list_of_records[mid][1],list_of_records[mid][2],list_of_records[mid][3])
        else:
            if list_of_records[mid][0] < search_id: #search_id is greater than the mid id
                return task1_4(list_of_records,search_id,high,mid+1)
            else:
                return task1_4(list_of_records,search_id,mid-1,low)
def main():
    product_id = input("Input a product ID: ")
    print(task1_4(task1_3(task1_1("TASK1.txt")),product_id,len(task1_3(task1_1("TASK1.txt")))-1,0))
    
main()
main()
```

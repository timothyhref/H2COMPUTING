```python
#task 2.1
class Tile:
    def __init__(self,symbol):
        self.symbol = symbol #symbol attribute
    def get_symbol(self):
        return self.symbol
class Treasure(Tile):
    def __init__(self,symbol,treasure):
        super().__init__(symbol)
        self.treasure = treasure
        
    def get_treasure(self):
        return self.treasure
class Bomb(Tile):
    def __init__(self,symbol,damage):
        super().__init__(symbol)
        self.damage = damage
    def get_damage(self):
        return self.damage
    
```

```python
#task 2.2
def display_grid(grid):
    for row in grid:
        for i in range(len(row)):
            if i != len(row)-1:
                print(row[i].symbol,end = " ") # space between objects
            else:
                print(row[i].symbol)
```

```python
#task 2.3
def main():
    grid = []
    for i in range(5):
        grid.append(['']*5)
        for j in range(5):
            grid[i][j] = Tile('.') #creates tile objects
    grid[1][2] = Treasure('T','Silver key')
    grid[2][4] = Treasure('T','Gold coin')
    grid[3][3] = Treasure('T','Jade pendant')
    grid[2][3] = Bomb('B',1)
    grid[0][0] = Tile('P')
    display_grid(grid)
    life = 1 #one life
    player_coord = [0,0] #stores player coord as a list
    inventory = [] #player's inventory
    movements = input("Please enter a sequence of moves (W/A/S/D):") #player input
    for i in range(len(movements)):
        item = movements[i]
        if item == 'W':
            if player_coord[0] - 1 >= 0: #check if in grid
                player_coord[0] += 1
        elif item == 'A':
            if player_coord[1] - 1 >= 0: #check if in grid
                player_coord[1] -= 1
        elif item == 'S':
            if player_coord[0] + 1 <= 4: #check if in grid
                player_coord[0] += 1
        else:
            if player_coord[1] + 1 <= 4: #check if in grid
                player_coord[1] += 1
        if grid[player_coord[0]][player_coord[1]].get_symbol() == "T":
            if grid[player_coord[0]][player_coord[1]].get_treasure not in inventory:
                inventory.append(grid[player_coord[0]][player_coord[1]].get_treasure()) #add treasure to inventory if not present
        elif grid[player_coord[0]][player_coord[1]].get_symbol() == "B":
            life -= grid[player_coord[0]][player_coord[1]].get_damage()
            if life - grid[player_coord[0]][player_coord[1]].get_damage() <= 0: #decrease life
                break
            
            
    if life <= 0:
        print("Game Over, no more lives.")
    elif len(inventory) == 3:
        print("Congratulations! You won the game!")
        print(f"Treasures collected {len(inventory)}/3: {inventory[0]}, {inventory[1]}, {inventory[2]}" )
    else:
        print("Sorry, you did not win the game.")
        collected = len(inventory)
        if collected == 0:
            print(f"Treasures collected {len(inventory)}/3: None" )
        elif collected == 1:
            print(f"Treasures collected {len(inventory)}/3: {inventory[0]}" )
        elif collected == 2:
            print(f"Treasures collected {len(inventory)}/3: {inventory[0]}, {inventory[1]}" )

#main
main() #win
main() #did not win
main() #game over
```

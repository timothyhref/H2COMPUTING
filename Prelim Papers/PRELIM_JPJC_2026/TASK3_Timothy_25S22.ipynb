```python
#task 3.1
def createSensorNode():
    import random
    id = random.randint(0,99999)
    id = str(id)
    if len(id) != 5:
        id = "0"*(5-len(id)) + id
    key = ["Left","SensorID","Right"]
    value = [None,id,None]
    return dict(zip(key,value))

print(createSensorNode())
```

```python
#task 3.2 (a)
def insert(Root,newSensorNode):
    if Root == None:
        return newSensorNode
    else:
        curr = Root
        while True:
            if newSensorNode.get('SensorID') > curr.get('SensorID'): #newnode sensor id is more than the curr
                if curr.get('Right') == None:
                    curr['Right'] = newSensorNode #set newnode sensor id to be the right child if free
                    break
                else:
                    curr = curr['Right'] 
            else: #newnode sensor id is less than the curr
                if curr.get('Left') == None:
                    curr['Left'] = newSensorNode #set newnode sensor id to be the left child if free
                    break
                else:
                    curr = curr['Left']
        return Root
```

```python
#task 3.2 (b)
def insertHandleDuplicate(Root,newSensorNode):
    if Root == None:
        return newSensorNode
    else:
        curr = Root
        while True:
            if newSensorNode.get('SensorID') == curr.get('SensorID'):
                newnode = createSensorNode() #create a new node
                return insertHandleDuplicate(Root,newnode) #get the root of the new insertion
            elif newSensorNode.get('SensorID') > curr.get('SensorID'): #newnode sensor id is more than the curr
                if curr.get('Right') == None:
                    curr['Right'] = newSensorNode #set newnode sensor id to be the right child if free
                    break
                else:
                    curr = curr['Right'] 
            else: #newnode sensor id is less than the curr
                if curr.get('Left') == None:
                    curr['Left'] = newSensorNode #set newnode sensor id to be the left child if free
                    break
                else:
                    curr = curr['Left']
        return Root
```

```python
#task 3.3
def inOrder(Root):
    if Root:
        inOrder(Root.get('Left')) #recusively goes down left branch
        print(Root.get('SensorID'),end=' ')
        inOrder(Root.get('Right')) #recusively goes down right branch
```

```python
#task 3.5
def isValidSensorID(sensorID):
    if len(sensorID) == 5 and sensorID.isdigit() == True:
        return True
    else:
        return False
```

```python
#task 3.6
#Test case 1
test_data_1 = "59346" #normal test type, len of 5 and contains multiple digits not the same that are between 0-9
#expected output: True
print(isValidSensorID(test_data_1))

#Test case 2
test_data_2 = "00000" #boundary test type, len of 5 but contains only 1 digit that is from between 0-9
#expected output: True
print(isValidSensorID(test_data_2))

#Test case 3
test_data_3 = "A12345" #erroneous test type, len of 6 and contains "A", an alphabet
#expected output: False
print(isValidSensorID(test_data_3))
```

```python
#Task 2.1
import random
def CreateIsland(rowsize,colsize):
    grid = []
    for i in range(rowsize):
        grid.append(['.']*colsize)
    grid[0][0] = "*"
    return grid

def plantCoconuts(grid):
    rowsize = len(grid)
    colsize = len(grid[0])
    for i in range(3):
        while True:
            row = random.randint(0,rowsize-1)
            col = random.randint(0,colsize-1)
            if grid[row][col] == ".":
                grid[row][col] = "C"
                break
    return grid

def display(grid):
    for row in grid:
        print(row)
#current[0] is the row/y current[1] is the col/x

def move(grid, current, direction, distance, coconut):
    initial = current.copy()
    if direction == "L":
        current[1] -= (distance)
        for i in range(1,distance):
            if grid[initial[0]][initial[1]-i] != "C":
                grid[initial[0]][initial[1]-i] = "*"
    elif direction == "R":
        current[1] += (distance)
        for i in range(1,distance):
            if grid[initial[0]][initial[1]+i] != "C":
                grid[initial[0]][initial[1]+i] = "*"
    elif direction == "U":
        current[0] -= (distance)
        for i in range(1,distance):
            if grid[initial[0]-i][initial[1]] != "C":
                grid[initial[0]-i][initial[1]] = "*"
    elif direction == "D":
        current[0] += (distance)
        for i in range(1,distance):
            if grid[initial[0]+i][initial[1]] != "C":
                grid[initial[0]+i][initial[1]] = "*"
    if grid[current[0]][current[1]] == "C":
        grid[current[0]][current[1]] = "E"
        coconut += 1
    else:
        grid[current[0]][current[1]] = "*"
    return (grid,current,coconut)
```

```python
def main():
    flag = False
    grid = CreateIsland(10,15)
    grid = plantCoconuts(grid)
    current = [0,0]
    coconut = 0
    display(grid)
    for i in range(8):
        print(f"You are at position: {current}\tCoconut eaten: {coconut}")
        direction, distance = input("Enter your move (L/R/U/D distance): ").split(" ")
        grid,current,coconut = move(grid,current,direction,int(distance),coconut)
        display(grid)
        if coconut == 3 and current == [9,14]:
            flag = True
            break
    if flag == True:
        print("You have won!\nYou have reached finish point and eaten 3 coconuts.")
    else:
        print("Sorry you did not win.")
        if current != [9,14]:
            print("You did not reach the finish point.")
        if coconut != 3:
            print("You have not eaten enough coconut.")
```

```python
main()
```

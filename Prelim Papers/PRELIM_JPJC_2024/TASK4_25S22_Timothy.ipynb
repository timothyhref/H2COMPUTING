```python
#Task 4.1
import sqlite3
conn = sqlite3.connect("Airport.db")
#flightNum is the primary key because its the unique identifier
conn.execute('''CREATE TABLE IF NOT EXISTS "Flight" (
	"flightNum"	TEXT NOT NULL,
	"departure"	TEXT NOT NULL,
	"destination"	TEXT NOT NULL,
	"departureTime"	TEXT NOT NULL,
	"arrivalTime"	TEXT NOT NULL,
	PRIMARY KEY("flightNum")
);''')
conn.commit()
conn.close()
```

```python
#Task 4.2
import sqlite3
conn = sqlite3.connect("Airport.db")
with open("TASK4.txt",'r') as file:
    lines = file.readlines()
lines = [lines[i].strip().split(",") for i in range(1,len(lines))] #split the comma seperated values in the text file into individual lists
for line in lines:
    conn.execute('''INSERT INTO Flight VALUES (?,?,?,?,?)''',(line)) #insert the values in the text into flight
conn.commit()
conn.close()
```

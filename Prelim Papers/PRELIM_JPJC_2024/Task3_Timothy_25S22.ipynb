```python
#Task 3.1
class Node:
    def __init__(self,Value):
        self.left = None
        self.right = None
        self.value = Value
```

```python
#Task 3.2
class Tree:
    def __init__(self,node):
        self.root = node
    def insert(self,node):
        curr = self.root
        while curr:
            if node.value < curr.value:
                if curr.left == None:
                    curr.left = node
                    return
                else:
                    curr = curr.left
            else:
                if curr.right == None:
                    curr.right = node
                    return
                else:
                    curr = curr.right
    def in_order_traversal(self,currNode):
        if currNode:
            self.in_order_traversal(currNode.left)
            print(currNode.value,end=" ")
            self.in_order_traversal(currNode.right)
    def pre_order_traversal(self,currNode):
        if currNode:
            print(currNode.value,end=" ")
            self.pre_order_traversal(currNode.left)
            self.pre_order_traversal(currNode.right)
    def flatten(self, currNode): #AI Generated Help
        if currNode is None: #Basecase, we reached the end of the subtree
            return None

        original_left = currNode.left #save copy of the original left
        original_right = currNode.right #save copy of the original right

        currNode.left = None 

        last_node = currNode #previous node

        if original_left: #if the current node's left is not none
            currNode.right = original_left #set the current node's right to the left of the current node (Pre-order traversal to LL)
            last_node = self.flatten(original_left) #recursively go down the left subtree
        else:
            currNode.right = None

        if original_right:
            last_node.right = original_right
            last_node = self.flatten(original_right)

        return last_node
    def ll_traversal(self,currNode):
        if currNode:
            print(currNode.value, end=" ")
            self.ll_traversal(currNode.right)
```

```python
#Task 3.3
BST = Tree(Node(5))
BST.insert(Node(3))
BST.insert(Node(2))
BST.insert(Node(4))
BST.insert(Node(7))
BST.insert(Node(8))
BST.in_order_traversal(BST.root)
print()
BST.pre_order_traversal(BST.root)
print()
BST.flatten(BST.root)
BST.ll_traversal(BST.root)
```

```python
#Task 1.1
def CharToNum(letter):
    return ord(letter)-65
```

```python
#Task 1.2
def encrypt(text,key):
    difference = (len(text)//len(key))+1
    keyf = (key*difference)[:len(text)]
    result = ""
    for i in range(len(text)):
        if text[i] == " ":
            result += "!"
            continue
        total = CharToNum(text[i]) + CharToNum(keyf[i])
        pos = total%26
        result += chr(pos+65)
    return result
```

```python
#Task 1.3
encrypted = []
with open("TASK1.txt","r") as file:
    lines = file.readlines()

with open("ENCRYPTED.txt","w") as file2:
    for line in lines:
        line.strip()
        encrypted.append(encrypt(line,"JPJC")+"\n")
    file2.writelines(encrypted)
    
```
